**Spotify Top 50 songs 2023 Analysis**

This project analyzes Spotify top tracks using various audio features
like popularity, energy, danceability, etc.

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from IPython.display import display
```

**Data Loading**

```python
df=pd.read_csv('/content/top_50_2023.csv')
display(df.head())
```

**Data Understanding**



In this section, the structure and basic information of the dataset will be given

```python
df.shape
df.columns
df.info()
df.describe()
```

**Data Cleaning**

Cleaning columns to ensure consistency and avoid errors

```python
df.columns = df.columns.str.strip().str.lower()
df.columns
```

**Top 10 Most Popular Songs**

Identifies top 10 popular songs based on popularity score

```python
top_songs=df.sort_values(by='popularity',ascending=False).head(10)
display(top_songs[['track_name','artist_name','popularity']])
```

**Top Artists**

Finding the artists who appear most frequently in the dataset

```python
display(pd.DataFrame(df['artist_name'].value_counts().head(10)))
```

**Explicit vs Non-Explicit Songs**

Analyzing the distribution of explicit and non-explicit songs

```python
display(pd.DataFrame(df['is_explicit'].value_counts()))
```

**Genre Analysis**

Identifying the most common genres in the dataset

```python
display(pd.DataFrame(df['genres'].value_counts().head(10)))
```

**Most Energetic Songs**

Finding songs with the highest energy levels

```python
energetic=df.sort_values(by='energy',ascending=False).head(10)
display(energetic[['track_name','artist_name','energy']])
```

**Most Danceable Songs**

Finding songs with the highest danceability scores

```python
danceable=df.sort_values(by='danceability',ascending=False).head(10)
display(danceable[['track_name','artist_name','danceability']])
```

**Top Songs Visualization**

sns.barplot(x='popularity',y='track_name',data=top_songs)
plt.title('Top 10 Most Popular Songs')
plt.show()

```python
sns.barplot(x='popularity',y='track_name',data=top_songs)
plt.title('Top 10 Most Popular Songs')
plt.show()
```

**Explicit Content Visualization**

Visualizing the distribution of explicit and non-explicit songs

```python
sns.countplot(x='is_explicit',data=df)
plt.title("Explicit vs non-Explicit Songs")
plt.show()
```

**Energy Distribution**

Analyzing the distribution of energy levels across songs

```python
sns.histplot(df['energy'],kde=True)
plt.title("Energy Distribution")
plt.show()
```

**Conclusion**

The analysis shows that popular songs tend to have higher energy and danceability. The dataset also reveals trends in genres and explicit content, reflecting listener preferences in modern music.
